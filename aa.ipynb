```python
def precision(relevant_retrieved, total_retrieved):
    """
    Precision = How many retrieved items are relevant
    """
    if total_retrieved == 0:
        return 0.0
    return relevant_retrieved / total_retrieved


def recall(relevant_retrieved, total_relevant):
    """
    Recall = How many relevant items were retrieved
    """
    if total_relevant == 0:
        return 0.0
    return relevant_retrieved / total_relevant


# Example usage
relevant_retrieved = 3
total_retrieved = 5
total_relevant = 4

print("Precision:", precision(relevant_retrieved, total_retrieved))
print("Recall:", recall(relevant_retrieved, total_relevant))
```

```python

from nltk.translate.bleu_score import sentence_bleu
from nltk.tokenize import word_tokenize

reference = "home loan requires identity proof and income documents"
candidate = "home loan needs identity proof and income proof"

reference_tokens = [word_tokenize(reference)]
candidate_tokens = word_tokenize(candidate)

bleu_score = sentence_bleu(reference_tokens, candidate_tokens)

print("BLEU Score:", bleu_score)
```

```python
from rouge_score import rouge_scorer

reference = "home loan documents include identity proof income proof and bank statements"
candidate = "documents required for home loan are identity proof and income proof"

scorer = rouge_scorer.RougeScorer(
    ['rouge1', 'rouge2', 'rougeL'],
    use_stemmer=True
)

scores = scorer.score(reference, candidate)

for metric, score in scores.items():
    print(metric, score)
```

```python
# Zero Division Error Handling
def zeroDivision():
    try:
        x = 10/0
    except ZeroDivisionError:
        print("cannot divide by 0")

zeroDivision()
```

```python
# Multiple Exception Error Handling
def mutliError():
    try:
        num = int(input())
        result = 10/num
        print(result)
    except ValueError:
        print("Invalid number format")
    except ZeroDivisionError:
        print("cannot divide by 0")
        
mutliError()
```

```python
# Generic except Exception
def GenericExcept():
    try:
        result = 10/2
    except Exception as e:
        print("Error Ocuured",e)
    else:
        print(result)  # else runs only if no exception occurs
GenericExcept()
```

```python
raw_logs = [ 
{"id": 1, "event": "  LOGIN_SUCCESS ", "level": "INFO", "user": "admin"}, 
{"id": 2, "event": "DATABASE_ERROR", "level": "CRITICAL", "user": "system"}, 
{"id": 3, "event": "  login_success ", "level": "info", "user": "user1"}, 
{"id": 4, "event": "TIMEOUT_API", "level": "WARNING", "user": "system"}, 
{"id": 5, "event": "database_error", "level": "critical", "user": "admin"}, 
]

def clean_data(data):
    result = []
    for row in data:
        new_log = row.copy()
        new_log["event"]= new_log["event"].strip().upper()
        new_log["level"]= new_log["level"].strip().upper()
        result.append(new_log)
    return result,data

clean_data(raw_logs)
```
